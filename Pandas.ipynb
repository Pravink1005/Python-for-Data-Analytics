# Pandas

Pandas is a Python library that helps us **read, organize, clean, analyze, and summarize data**. It is commonly used when working with Excel files, CSV files, and datasets.

## What, Why, and Where?

- **What is Pandas?** A Python library for working with structured data in tables.
- **Why use it?** It makes filtering, cleaning, calculating, and summarizing data easier.
- **Where is it used?** Data analysis, business reports, dashboards, and machine-learning preparation.


## 1. Install and import Pandas

We usually import it using the short name `pd`.

```python
import pandas as pd

print("Pandas version:", pd.__version__)
```

## 2. Create your first Series

A **Series** is like one column in a table.

```python
marks = pd.Series([85, 90, 78, 92])
print(marks)

# Print without index and dtype
print(marks.to_string(index=False))
```

**Understand it:** Each value has an index (0, 1, 2, 3). Pandas creates these index labels automatically.

**Try it:** Create a Series containing five prices.

```python
prices = pd.Series([120, 250, 80, 500, 175])
prices
```

## 3. Create a DataFrame

A **DataFrame** is like a table with rows and columns.

```python
data = {
    "Name": ["Anu", "Bala", "Charan", "Divya", "Ezhil"],
    "Age": [21, 22, 20, 23, 21],
    "City": ["Chennai", "Coimbatore", "Madurai", "Chennai", "Salem"],
    "Marks": [85, 72, 91, 66, 78],
    "Attendance": [92, 85, 96, 70, 88]
}

df = pd.DataFrame(data)
df

# print(df.to_string(index=False))
```

**What happened?** We made a table from a Python dictionary. Each dictionary key became a column name.

- **Row:** information about one student.
- **Column:** one type of information, such as `Marks`.
- **DataFrame:** the complete table.

## 4. View the data

These commands help you quickly understand a dataset.

```python
print("First 3 rows:")
display(df.head(3))

print("Last 2 rows:")
display(df.tail(2))

print("Dataset size (rows, columns):", df.shape)
print("Column names:", df.columns.tolist())
print("Column data types:")
df.dtypes
```

- `head(n)` shows the first *n* rows.
- `tail(n)` shows the last *n* rows.
- `shape` gives `(number of rows, number of columns)`.
- `dtypes` shows the type of data in each column.

## 5. Select columns and rows

Use a column name to select one column. Use `loc` for label-based selection and `iloc` for position-based selection.

```python
# One column (returns a Series)
df["Name"]
```

```python
# Multiple columns (returns a DataFrame)
df[["Name", "Marks"]]
```

```python
# Rows with index labels 0 and 1
df.loc[0]
```

```python
# First two rows by position
df.iloc[1:2]
```

**Remember:**
- `df["Marks"]` selects the Marks column.
- `df[["Name", "Marks"]]` selects two columns.
- `loc` uses index labels; `iloc` uses integer positions.
- Python slices usually exclude the ending position, but `loc[0:1]` includes label `1`.

## 6. Filter rows

Filtering means keeping only rows that match a condition.

```python
# Students who scored at least 80
high_marks = df[df["Marks"] >= 80]
high_marks
```

```python
# Students from Chennai
chennai_students = df[df["City"] == "Chennai"]
chennai_students
```

```python
# Students with marks >= 80 AND attendance >= 90
df[(df["Marks"] >= 80) & (df["Attendance"] >= 90)]
```

**Tip:** Use `&` for AND and `|` for OR. Put each condition inside parentheses.

## 9. Basic calculations and summary statistics

Pandas can quickly calculate totals, averages, minimums, and maximums.

```python
print("Total marks:", df["Marks"].sum())
print("Average marks:", df["Marks"].mean())
print("Highest marks:", df["Marks"].max())
print("Lowest marks:", df["Marks"].min())
print("Median marks:", df["Marks"].median())
```

## 11. Group data with `groupby()`

`groupby()` lets us summarize data by category—for example, average marks by city.

```python
# Average marks for each city
df.groupby("City")["Marks"].mean().sort_values(ascending=False)
```

```python
# Several summaries for each city
df.groupby("City").agg(
    Student_Count=("Name", "count"),
    Average_Marks=("Marks", "mean"),
    Highest_Marks=("Marks", "max")
).round(2)
```

**Simple idea:** Split rows into groups, calculate something for each group, and show the results.

## 12. Find missing values

Real datasets often contain empty cells. Pandas represents many missing values as `NaN`.

```python
messy_data = pd.DataFrame({
    "Name": ["Arun", "Banu", "Cathy", "Dinesh"],
    "Age": [22, None, 21, 24],
    "Marks": [80, 75, None, 90],
    "City": ["Chennai", "Madurai", None, "Salem"]
})

messy_data
```

```python
# Count missing values in each column
messy_data.isna().sum()
```

### Ways to handle missing values

There is no single best solution. Choose based on what the data means.

```python
# # Option A: remove rows that contain any missing value
# complete_rows = messy_data.dropna()
# display(complete_rows)

# Option B: fill missing numeric values with the column average
filled_data = messy_data.copy()
filled_data["Age"] = filled_data["Age"].fillna(filled_data["Age"].mean())
filled_data["Marks"] = filled_data["Marks"].fillna(filled_data["Marks"].mean())
filled_data["City"] = filled_data["City"].fillna("Unknown")
filled_data
```

**Important:** Filling with an average or `"Unknown"` is only an example. Before cleaning real data, investigate why values are missing and choose a sensible approach.

## 13. Find and remove duplicate rows

```python
duplicate_data = pd.DataFrame({
    "Name": ["Anu", "Bala", "Anu", "Charan"],
    "Marks": [85, 72, 85, 91]
})

print("Duplicate row flags based on Name and Marks:")
display(duplicate_data.duplicated(subset=['Name', 'Marks']))

# Remove duplicates looking at Name and Marks specifically
clean_duplicates = duplicate_data.drop_duplicates(subset=['Name', 'Marks'])

print("\nCleaned DataFrame:")
print(clean_duplicates)
```

- `duplicated()` marks repeated rows as `True`.
- `drop_duplicates()` returns a DataFrame with duplicate rows removed.
- If only certain columns define a duplicate, use `drop_duplicates(subset=["Name"])`—but confirm that rule makes sense for your data.

## 14. Work with text

String methods help clean and search text values.

```python
names_data = pd.DataFrame({
    "Name": [" anu ", "BALA", "charan ", "Divya"],
    "Department": [" data ", "HR", "DATA", "Finance"]
})

names_data["Name_Clean"] = names_data["Name"].str.strip().str.title()
names_data["Department_Clean"] = names_data["Department"].str.strip().str.upper()
names_data
```

- `str.strip()` removes extra spaces at the start and end.
- `str.title()` capitalizes words.
- `str.upper()` converts text to uppercase.

## 15. Read a CSV file

CSV files are common in data analytics. `read_csv()` loads one into a DataFrame.

```python
# Create a small example CSV, then read it back.
sample_csv = pd.DataFrame({
    "Product": ["Pen", "Notebook", "Bag"],
    "Price": [10, 50, 600],
    "Quantity": [20, 12, 3]
})

sample_csv.to_csv("sample_sales.csv", index=False)
sales = pd.read_csv("sample_sales.csv")
sales
```

## 17. Merge two DataFrames

A merge combines tables using a shared column, similar to a SQL join.

```python
students = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Name": ["Anu", "Bala", "Charan"]
})

scores = pd.DataFrame({
    "Student_ID": [1, 2, 3],
    "Marks": [85, 72, 91]
})

merged = pd.merge(students, scores, on="Student_ID", how="inner")
merged
```

- `on="Student_ID"` says which column matches.
- `how="inner"` keeps only IDs found in both tables.
- Other common options are `left`, `right`, and `outer`.
