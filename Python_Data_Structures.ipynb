## 11. List Comprehension

List comprehension is a short way to create a new list.

### Normal approach


````python
numbers = [1, 2, 3, 4, 5]
squares = []

for number in numbers:
    squares.append(number ** 2)

print(squares)
````

### List comprehension approach

````python
numbers = [1, 2, 3, 4, 5]

squares = [number ** 2 for number in numbers]

print(squares)
````

````python
# List comprehension with a condition
numbers = [10, 15, 20, 25, 30]

even_numbers = [n for n in numbers if n % 2 == 0]

print(even_numbers)
````

# 14. Sets

### What?
A set is a collection of **unique** values.

### Why?
Sets are useful when you want to remove duplicates or compare groups of values.


````python
numbers = {10, 20, 20, 30, 30, 40}

print(numbers)
````

### Set Operations

````python
A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

print("Union:", A | B)
print("Intersection:", A & B)
print("Difference A-B:", A - B)
print("Difference B-A:", B - A)
````

### Meaning

- **Union** → everything from both sets
- **Intersection** → values common to both
- **Difference** → values present in one set but not the other

# 26. Mutable vs Immutable

### Mutable
The object can be changed after creation.

Examples:
- List
- Dictionary
- Set

### Immutable
The object cannot be changed after creation.

Examples:
- Tuple
- String
- Integer



````python
# List is mutable
numbers = [10, 20, 30]
numbers[0] = 100

print(numbers)
````

````python
# Tuple is immutable
numbers = (10, 20, 30)

# Uncommenting the next line will produce an error:
# numbers[0] = 100
````

### Simple memory trick

**List = Can change**

**Tuple = Cannot change**

This matters when designing reliable data structures and working with data that should remain fixed.


# 27. Data Analysis Practice — Sales

Now we will combine lists, loops and basic functions to analyze sales data.


````python
sales = [1200, 1500, 900, 2100, 1800, 2500]

total_sales = sum(sales)
average_sales = total_sales / len(sales)
highest_sales = max(sales)
lowest_sales = min(sales)

print("Total Sales:", total_sales)
print("Average Sales:", average_sales)
print("Highest Sale:", highest_sales)
print("Lowest Sale:", lowest_sales)
````

### Find sales above the average

````python
sales = [1200, 1500, 900, 2100, 1800, 2500]

average_sales = sum(sales) / len(sales)

above_average = [sale for sale in sales if sale > average_sales]

print("Average:", average_sales)
print("Above average:", above_average)
````

# 28. Employee Data Analysis

Real datasets often contain multiple records. Let's use a list of dictionaries.


````python
employees = [
    {"name": "Arun", "department": "IT", "salary": 30000},
    {"name": "Bala", "department": "HR", "salary": 35000},
    {"name": "Cathy", "department": "IT", "salary": 45000},
    {"name": "Deepak", "department": "Finance", "salary": 40000}
]

# Total salary
total_salary = sum(employee["salary"] for employee in employees)

# Average salary
average_salary = total_salary / len(employees)

print("Total Salary:", total_salary)
print("Average Salary:", average_salary)
````

### Find employees with salary above average

````python
above_average = [
    employee for employee in employees
    if employee["salary"] > average_salary
]

for employee in above_average:
    print(employee)
````

# 29. Mini Project — Student Marks Analyzer

## Goal

Build a small program that analyzes student marks.

Each student has:

- Name
- Marks in 3 subjects

Your program should calculate:

1. Total marks
2. Average marks
3. Highest average
4. Lowest average
5. Pass/Fail
6. Students with average >= 75

### Dataset


````python
students = [
    {"name": "Arun", "math": 80, "science": 75, "english": 85},
    {"name": "Bala", "math": 65, "science": 70, "english": 60},
    {"name": "Cathy", "math": 90, "science": 88, "english": 95},
    {"name": "Deepak", "math": 72, "science": 68, "english": 75}
]

for student in students:
    total = student["math"] + student["science"] + student["english"]
    average = total / 3

    print(
        student["name"],
        "Total:", total,
        "Average:", round(average, 2)
    )
````

### Step 1 — Add Total and Average

````python
students = [
    {"name": "Arun", "math": 80, "science": 75, "english": 85},
    {"name": "Bala", "math": 65, "science": 70, "english": 60},
    {"name": "Cathy", "math": 90, "science": 88, "english": 95},
    {"name": "Deepak", "math": 72, "science": 68, "english": 75}
]

for student in students:
    student["total"] = student["math"] + student["science"] + student["english"]
    student["average"] = student["total"] / 3

print(students)
````

### Step 2 — Add Pass/Fail
Assume a student passes when the average is at least 50.

````python
for student in students:
    if student["average"] >= 50:
        student["result"] = "Pass"
    else:
        student["result"] = "Fail"

for student in students:
    print(student["name"], student["average"], student["result"])
````

### Step 3 — Find Highest and Lowest Average

````python
highest_student = max(students, key=lambda student: student["average"])
lowest_student = min(students, key=lambda student: student["average"])

print("Highest:", highest_student["name"], highest_student["average"])
print("Lowest:", lowest_student["name"], lowest_student["average"])
````

### Step 4 — Find Students With Average >= 75

````python
top_students = [
    student for student in students
    if student["average"] >= 75
]

for student in top_students:
    print(student["name"], student["average"])
````

### Step 5 — Final Report

````python
print("=" * 50)
print("STUDENT MARKS REPORT")
print("=" * 50)

for student in students:
    print(
        f'{student["name"]}: '
        f'Total={student["total"]}, '
        f'Average={student["average"]:.2f}, '
        f'Result={student["result"]}'
    )
````
