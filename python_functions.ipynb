> A function is a reusable block of code that performs a task.


## 1. What is a function?

A **function** is a named block of code that you can run whenever you need it.

### Why do we use functions?
- **Reuse:** write code once and use it many times.
- **Organize:** split a large program into smaller tasks.
- **Readability:** give a task a meaningful name.
- **Maintenance:** update the code in one place.

Without a function, you might repeat the same calculation many times. A function lets you reuse it.


```python
# A simple function
def say_hello():
    print("Hello! Welcome to Python.")

# Calling (running) the function
say_hello()
say_hello()  # We can reuse it
```

## 2. Creating a function with `def` and calling it

- `def` tells Python that you are defining a function.
- The function name should describe its task.
- The indented lines belong to the function.
- Defining a function does not run it. You call it using parentheses: `function_name()`.


```python
def show_message():
    print("I am learning Python functions.")

show_message()  # Function call
```

## 3. Parameters and arguments

A **parameter** is a variable written inside the function definition. An **argument** is the actual value passed when you call the function.

In `def greet(name):`, `name` is a parameter. In `greet("Nanthini")`, `"Nanthini"` is an argument.


```python
def greet(name):  # name is a parameter
    print("Hello,", name)

greet("Nanthini")  # "Nanthini" is an argument
greet("Ammu")
```

## 4. Multiple parameters

A function can accept more than one parameter. Pass values in the matching order when using positional arguments.


```python
def add_numbers(a, b):
    print("Sum:", a + b)

add_numbers(10, 5)
add_numbers(100, 25)
```

## 5. `return` and the difference between `print()` and `return`

- `print()` displays something on the screen.
- `return` sends a result back to the code that called the function. You can store or reuse that result.
- A function without an explicit `return` gives back `None`.


```python
def add_and_print(a, b):
    print(a + b)  # Displays the answer

def add_and_return(a, b):
    return a + b  # Sends the answer back

# add_and_print(3, 4)

result = add_and_return(3, 4)
print("Stored result:", result)
print("We can reuse it:", result * 2)
```

## 6. Default arguments

A **default argument** is used when the caller does not provide a value for that parameter.


```python
def greet_person(name, greeting="Hello"):
    print(greeting, name)

# greet_person("Nanthini")              # Uses the default greeting
greet_person("Good morning","Ammu")  # Uses the supplied greeting
```

## 7. Positional arguments and keyword arguments

- **Positional arguments:** values are matched to parameters by their position.
- **Keyword arguments:** specify the parameter name, so the order can be changed.
- In a call, positional arguments must come before keyword arguments.


```python
def employee_info(name, role, salary):
    print(f"{name} works as {role} and earns Rs. {salary}.")

# Positional arguments
employee_info("Ravi", "Analyst", 30000)

# Keyword arguments (order can change)
employee_info(salary=35000, name="Meena", role="Developer")
```

## 8. `*args` — a variable number of positional arguments

Use `*args` when you want a function to accept any number of positional arguments. Inside the function, `args` is a tuple.


```python
def total_marks(*args):
    print("Values received:", args)
    return sum(args)

print(total_marks(80, 90))
print(total_marks(80, 90, 75, 88))
```

## 9. `**kwargs` — a variable number of keyword arguments

Use `**kwargs` when you want to accept any number of named arguments. Inside the function, `kwargs` is a dictionary.


```python
def show_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")

show_details(name="Nanthini", role="Data Analyst", city="Coimbatore",age=23)
```

## 10. Local vs global variables

- A **local variable** is created inside a function and is normally usable only inside it.
- A **global variable** is created outside functions and can be read inside a function.
- Prefer passing values as parameters and returning results rather than changing globals unnecessarily.


```python
company = "ABC Analytics"  # Global variable

def show_company():
    employee_count = 25  # Local variable
    print("Company:", company)
    print("Employees:", employee_count)

    print(employee_count)
show_company()
# print(employee_count)  # Uncommenting this causes NameError: it is local
```

## 11. Functions with lists

Lists can be passed to functions just like other values. A function can calculate or process the list.


```python
def calculate_average(numbers):
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

scores = [70, 85, 90, 65]
print("Scores:", scores)
print("Average:", calculate_average(scores))
```

## 12. Functions with dictionaries

A dictionary stores key-value pairs. Use `.get()` to safely read a key that might not exist.


```python
def display_employee(employee):
    print("Name:", employee.get("name", "Unknown"))
    print("Department:", employee.get("department", "Not specified"))
    print("Salary:", employee.get("salary", 0))

employee = {"name": "Priya", "department": "Data", "salary": 32000}
display_employee(employee)
```

## 13. Functions with loops

A function can use a loop to process each item in a collection.


```python
def print_each_name(names):
    for name in names:
        print(name)

print_each_name(["Nanthini", "Ammu", "Kayal"])
```

## 14. Functions with conditions

Use `if`, `elif`, and `else` inside a function to make decisions.


```python
def salary_level(salary):
    if salary >= 50000:
        return "High"
    elif salary >= 30000:
        return "Medium"
    else:
        return "Entry-level"

print(salary_level(60000))
print(salary_level(40000))
print(salary_level(22000))
```

## 15. Functions calling other functions

One function can call another function. This helps split a bigger task into small, understandable steps.


```python
def calculate_bonus(salary):
    return salary * 10

def create_salary_summary(name, salary):
    bonus = calculate_bonus(salary)
    total = salary + bonus
    return f"{name}: salary Rs. {salary}, bonus Rs. {bonus}, total Rs. {total}"

print(create_salary_summary("Arun", 40000))
```

## 16. Lambda functions

A **lambda** is a short anonymous function, often used for a small one-line operation.

Syntax: `lambda parameters: expression`

Use a regular `def` function when the logic is longer or needs multiple statements.


```python
# Regular function
def square(number):
    return number * number

# Equivalent short lambda
square_lambda = lambda number: number * number

print(square(5))
print(square_lambda(5))
```

## 17. `map()` — apply a function to every item

`map(function, iterable)` applies a function to each item. In Python 3, convert its result to `list()` to see all values at once.


```python
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda number: number ** 2, numbers))
print(squares)
```

## 18. `filter()` — keep items that match a condition

`filter(function, iterable)` keeps items for which the function returns `True`.


```python
numbers = [12, 7, 20, 3, 18, 9]
large_numbers = list(filter(lambda number: number >= 10, numbers))
print(large_numbers)
```

## 19. `sorted()` with functions and lambda

`sorted()` returns a new sorted list. Use `key=` to decide what value Python should sort by. Use `reverse=True` for descending order.


```python
employees = [
    {"name": "Ravi", "salary": 28000},
    {"name": "Meena", "salary": 42000},
    {"name": "Arun", "salary": 35000}
]

# Sort by salary, low to high
by_salary = sorted(employees, key=lambda employee: employee["salary"])
print(by_salary)

# Sort by salary, high to low
by_salary_high_to_low = sorted(
    employees, key=lambda employee: employee["salary"], reverse=True
)
print(by_salary_high_to_low)
```

## 20. Practical data-analysis functions

Functions make repeated data-cleaning and summary tasks easier to reuse. These examples use only Python's built-in features.


```python
def clean_names(names):
    # Remove spaces at the ends and skip blank names
    return [name.strip().title() for name in names if name.strip()]

def average(values):
    return sum(values) / len(values) if values else 0

def count_above(values, threshold):
    return len([value for value in values if value > threshold])

raw_names = [" nanthini ", "AMMU", "", "  kayal"]
sales = [1200, 1800, 900, 2200, 1500]

print("Clean names:", clean_names(raw_names))
print("Average sales:", average(sales))
print("Sales above 1500:", count_above(sales, 1500))
```

## 21. Mini Project — Employee Salary Analyzer

### Goal
Create reusable functions to:
1. Calculate average salary.
2. Find employees above a salary threshold.
3. Find the highest-paid employee.
4. Add a 10% bonus.
5. Display a simple report.

Run the cell and then try changing the employee data.


```python
employees = [
    {"name": "Nanthini", "department": "Data", "salary": 28000},
    {"name": "Ammu", "department": "HR", "salary": 32000},
    {"name": "Kayal", "department": "Data", "salary": 45000},
    {"name": "Ravi", "department": "IT", "salary": 52000},
    {"name": "Priya", "department": "IT", "salary": 38000}
]

def get_salaries(employee_list):
    return [employee["salary"] for employee in employee_list]

def average_salary(employee_list):
    salaries = get_salaries(employee_list)
    return sum(salaries) / len(salaries) if salaries else 0

def employees_above(employee_list, threshold):
    return [
        employee for employee in employee_list
        if employee["salary"] > threshold
    ]

def highest_paid(employee_list):
    if not employee_list:
        return None
    return max(employee_list, key=lambda employee: employee["salary"])

def salary_with_bonus(employee, bonus_rate=0.10):
    return employee["salary"] * (1 + bonus_rate)

def print_report(employee_list):
    if not employee_list:
        print("No employee data available.")
        return

    print("EMPLOYEE SALARY REPORT")
    print("-" * 25)
    print("Number of employees:", len(employee_list))
    print(f"Average salary: Rs. {average_salary(employee_list):,.2f}")

    top_employee = highest_paid(employee_list)
    print(f"Highest paid: {top_employee['name']} (Rs. {top_employee['salary']:,})")

    print("\nEmployees earning more than Rs. 35000:")
    for employee in employees_above(employee_list, 35000):
        print(f"- {employee['name']}: Rs. {employee['salary']:,}")

    print("\nEstimated salary including 10% bonus:")
    for employee in employee_list:
        total = salary_with_bonus(employee)
        print(f"- {employee['name']}: Rs. {total:,.2f}")

print_report(employees)
```
