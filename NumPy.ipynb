# NumPy for Data Analysis 

**Goal:** Learn the NumPy basics used in data analysis through small examples.

NumPy is a Python library that helps us work with numbers and arrays quickly. It is commonly used for calculations, statistics, and preparing data for tools such as pandas and machine-learning libraries.

## 1. Import NumPy

We usually import NumPy with the short name `np`. This makes its functions easier to write.

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 2. Create a NumPy array

An **array** stores multiple values together. It is similar to a Python list have can mix strings, integers, objects. In Numpy all elements must be the same dtype

```python
daily_sales = np.array([120, 150, 90, 200, 175])

print("Daily sales:", daily_sales)
print("Type:", type(daily_sales))
```

## 3. Python list vs NumPy array

With a Python list, `+` joins lists. With a NumPy array, `+` adds matching values when the arrays have the same shape. Join a Two Array using concatenate()

```python
python_list = [1, 2, 3]
numpy_array = np.array([1, 2, 3])

print("List + list:", python_list + python_list)
print("Array + array:", numpy_array + numpy_array)

joined_array = np.concatenate((numpy_array, numpy_array))
print("Joined array:", joined_array) 
```

## 4. Check the shape, size, and dimensions

- `shape`: how many values are in each dimension.
- `size`: total number of values.
- `ndim`: number of dimensions.

```python
scores = np.array([78, 85, 92, 88])

print("Scores:", scores)
print("Shape:", scores.shape)
print("Total values:", scores.size)
print("Dimensions:", scores.ndim)
```

## 5. Create arrays quickly

NumPy has helpful functions for creating arrays of zeros, ones, sequences, and evenly spaced values.

```python
print("Zeros:", np.zeros(4))
print("Ones:", np.ones(3))
print("Numbers from 0 to 5:", np.arange(6))
print("Even numbers from 2 to 10:", np.arange(2, 11, 2))
print("Five values from 0 to 1:", np.linspace(0, 1, 5))
```

## 6. Select values using indexing

Python uses zero-based indexing: the first item is at index `0`. Negative index `-1` means the last item.

```python
temperatures = np.array([29, 31, 30, 33, 32])

print("First temperature:", temperatures[0])
print("Third temperature:", temperatures[2])
print("Last temperature:", temperatures[-1])
```

## 7. Select a range using slicing

`array[start:stop]` includes `start` but does not include `stop`.

```python
numbers = np.array([10, 20, 30, 40, 50, 60])

print("Index 1 to 3:", numbers[1:4])
print("First three:", numbers[:3])
print("From index 3 onward:", numbers[3:])
```

## 8. Do calculations on every value

NumPy performs element-wise calculations. For example, adding `5` adds five to every value.

```python
prices = np.array([100, 200, 300])

print("Original prices:", prices)
print("Add 10:", prices + 10)
print("Discount of 10%:", prices * 0.90)
print("Double prices:", prices * 2)
print("Prices divided by 2:", prices / 2)
```

## 9. Filter data with a condition

A condition such as `scores >= 80` creates True/False values. Using that condition inside the array selects only matching values.

```python
exam_scores = np.array([45, 67, 82, 91, 76, 88])

passed_scores = exam_scores[exam_scores >= 50]
high_scores = exam_scores[exam_scores >= 80]

print("All scores:", exam_scores)
print("Scores at least 50:", passed_scores)
print("Scores at least 80:", high_scores)
```

## 10. Basic statistics

These functions help summarize data. The mean is the average; the median is the middle value after sorting; standard deviation describes how spread out the values are.

```python
monthly_sales = np.array([1200, 1500, 1100, 1800, 1400])

print("Total:", np.sum(monthly_sales))
print("Mean (average):", np.mean(monthly_sales))
print("Median:", np.median(monthly_sales))
print("Minimum:", np.min(monthly_sales))
print("Maximum:", np.max(monthly_sales))
print("Standard deviation:", round(np.std(monthly_sales), 2))
```

## 11. Work with a 2D array

A 2D array looks like a table with rows and columns. Here, each row represents a student and each column represents a subject.

```python
marks = np.array([
    [80, 75, 90],
    [65, 88, 70],
    [92, 95, 89]
])

print("Marks table:\n", marks)
print("Shape (rows, columns):", marks.shape)
print("First student's marks:", marks[0])
print("Marks in the first subject:", marks[:, 0])
print("Value in row 2, column 3:", marks[1, 2])
```

## 12. Calculate totals and averages by axis

For a 2D array, `axis=1` works across each row; `axis=0` works down each column.

```python
print("Total marks per student:", np.sum(marks, axis=1))
print("Average marks per student:", np.mean(marks, axis=1))
print("Average per subject:", np.mean(marks, axis=0))
```

## 13. Reshape an array

`reshape()` changes the layout without changing the number of values. Six values can become 2 rows by 3 columns.

```python
values = np.arange(1, 7)
table = values.reshape(2, 3)

print("Original:", values)
print("Reshaped table:\n", table)
```

## 14. Handle missing numerical values (`NaN`)

`NaN` means “Not a Number” and is often used to represent missing numerical data. Ordinary `mean()` returns `NaN` if a value is missing; `nanmean()` ignores NaN values.

```python
measurements = np.array([10.0, 12.0, np.nan, 14.0])

print("Measurements:", measurements)
print("Mean including NaN:", np.mean(measurements))
print("Mean ignoring NaN:", np.nanmean(measurements))
```

## 15. Small data-analysis example: weekly shop sales

We will analyze sales for one week and identify days with sales above 1500.

```python
days = np.array(["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"])
sales = np.array([1200, 1600, 1450, 1900, 1750, 2300, 2100])

average_sales = np.mean(sales)
best_day_index = np.argmax(sales)
above_average_days = days[sales > average_sales]

print("Average daily sales:", round(average_sales, 2))
print("Highest sales:", sales[best_day_index])
print("Best sales day:", days[best_day_index])
print("Days above average:", above_average_days)
```

## 18. Quick revision

| Function / feature | Meaning |
|---|---|
| `np.array([...])` | Create an array |
| `array.shape` | Dimensions by axis |
| `array.size` | Total number of values |
| `array.ndim` | Number of dimensions |
| `array[start:stop]` | Slice a range |
| `np.sum(array)` | Add values |
| `np.mean(array)` | Calculate average |
| `np.median(array)` | Calculate median |
| `np.min(array)` / `np.max(array)` | Smallest / largest value |
| `np.argmax(array)` | Index of the largest value |
| `np.arange(start, stop, step)` | Create a number sequence |
| `np.reshape(...)` | Change array shape |
| `np.nanmean(array)` | Average while ignoring NaN |

**Main idea:** NumPy helps you store numerical data in arrays, calculate across many values, filter values by conditions, and summarize datasets efficiently.
