# 4. Data Types

## What?

A data type tells Python **what kind of value something is**.

Python needs to know how to work with a value.

````python
age = 25
salary = 35000.50
name = "Pravin"
is_working = True

print(type(age))
print(type(salary))
print(type(name))
print(type(is_working))
````

Type conversion means changing one data type into another.


## Where?

Type conversion is important in:

- CSV files


# 9. Logical Operators

Logical operators are very important when filtering and analyzing data.

````python
age = 25
salary = 40000

print(age >= 18 and salary >= 30000)
````

### `or`


````python
age = 17
has_permission = True

print(age >= 18 or has_permission)
````

# 10. Conditional Statements

## What?

Conditional statements allow Python to **make decisions**.

Main keywords:

```python

```

## Why?

Programs need to behave differently depending on conditions.

## Where?

They are extremely important in:

- Data cleaning

````python
age = 20

if age >= 18:
    print("Adult")
````

### `if / else`


````python
age = 16

if age >= 18:
    print("Adult")
else:
    print("Minor")
````

````python
marks = 78

if marks >= 90:
    print("A")
elif marks >= 75:
    print("B")
elif marks >= 50:
    print("C")
else:
    print("Fail")
````

# 13. `break` and `continue`

## `break`

### What?

Stops the loop completely.

### Why?

Use it when you have reached the point where you no longer need the loop.

### Example


````python
for number in range(1, 11):
    if number == 5:
        break

    print(number)
````

## `continue`

### What?

Skips the current iteration and continues the loop.

### Example


````python
for number in range(1, 6):
    if number == 3:
        continue

    print(number)
````

## Mini Project — Solution

Use this only after attempting the problem yourself.


````python
employee_name = "Pravin"
salary = 40000
experience = 2

yearly_salary = salary * 12

if experience >= 3:
    bonus_rate = 0.20
elif experience >= 1:
    bonus_rate = 0.10
else:
    bonus_rate = 0.05

bonus = yearly_salary * bonus_rate
final_salary = yearly_salary + bonus

print("Employee:", employee_name)
print("Monthly Salary:", salary)
print("Experience:", experience, "years")
print("Bonus:", bonus_rate * 100, "%")
print("Yearly Salary:", yearly_salary)
print("Final Salary:", final_salary)
print("Above ₹50,000:", final_salary > 50000)
````
