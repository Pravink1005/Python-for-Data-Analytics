## 1. File Handling

### What is file handling?
File handling means creating, reading, updating, and saving information in a file using Python.

## 2. Writing to a text file

Use `open()` to open a file. The mode `"w"` means **write**.

- If the file does not exist, Python creates it.
- If it already exists, `"w"` replaces its old content.

The `with` statement closes the file automatically when the block finishes.

````python
with open("sample.txt", "w", encoding="utf-8") as file:
    file.write("Hello, Python!\n")
    file.write("I am learning file handling.")

print("File saved successfully!")
````

### Read the file

Mode `"r"` means **read**. The file must already exist.

````python
with open("sample.txt", "r", encoding="utf-8") as file:
    content = file.read()

print(content)
````

## 3. Reading a file line by line

`readline()` reads one line. Looping through the file lets us process each line separately.

````python
with open("sample.txt", "r", encoding="utf-8") as file:
    for line in file:
        print(line.strip())  # strip() removes extra spaces and the newline
````

## 4. Adding content to a file

Mode `"a"` means **append**. It adds new content at the end without deleting the existing content.

````python
with open("sample.txt", "a", encoding="utf-8") as file:
    file.write("\nThis line was added later.")

with open("sample.txt", "r", encoding="utf-8") as file:
    print(file.read())
````

## 5. File modes quick guide

| Mode | Meaning | Important note |
|---|---|---|
| `"r"` | Read | File should exist |
| `"w"` | Write | Creates or overwrites a file |
| `"a"` | Append | Adds to the end of a file |
| `"x"` | Create | Fails if the file already exists |
| `"r+"` | Read and write | File should exist |

**Tip:** Use `with open(...)` so Python closes the file for you.

## 6. Handling errors with `try` and `except`

If a file does not exist, reading it can raise `FileNotFoundError`. We can handle that error gracefully.

````python
try:
    with open("missing_file.txt", "r", encoding="utf-8") as file:
        print(file.read())
except FileNotFoundError:
    print("Sorry, that file does not exist yet.")
````

# Part 2 — CSV Files

## 7. What is a CSV file?

CSV means **Comma-Separated Values**. It stores table-like data as rows of text.

Example:

```text
name,age,city
Anu,20,Chennai
Ravi,22,Coimbatore
```

### Why and where is CSV useful?
CSV files are commonly used to exchange data between Excel, databases, Python programs, and data analysis tools.

Python has a built-in module called `csv`, so you do not need to install anything.

## 8. Writing a CSV file

The `csv.writer()` function helps write rows. `newline=""` helps avoid extra blank lines on some systems.

````python
import csv

students = [
    ["name", "age", "city"],
    ["Anu", 20, "Chennai"],
    ["Ravi", 22, "Coimbatore"],
    ["Meena", 21, "Madurai"]
]

with open("students.csv", "w", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerows(students)

print("CSV file created!")
````

## 9. Reading a CSV file

`csv.reader()` reads the file row by row. Each row is returned as a list of strings.

````python
import csv

with open("students.csv", "r", newline="", encoding="utf-8") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
````

# Modules

## 10. What is a module?

A **module** is a Python file containing reusable code, such as functions or variables.

### Why do we use modules?
- Avoid repeating code
- Keep programs organised
- Reuse functions in different programs

### Where are modules useful?
For example, a data analysis project might keep cleaning functions in `cleaning.py` and import them into the main program.

## 11. Using a built-in module

Python includes modules such as `math`, `random`, and `datetime`. Import a module to use its features.

````python
import math

print(math.sqrt(25))  # square root of 25
print(math.pi)        # value of pi
````

You can import only one function if you prefer.

````python
from math import sqrt

print(sqrt(81))
print(math.pi)
````

````python
import random
# 1. Get a random integer between 1 and 10
number = random.randint(1, 10)
print(f"Random Number: {number}")
````

````python
# 2. Pick a random item from a list
fruits = ["apple", "banana", "cherry", "date"]
picked_fruit = random.choice(fruits)
print(f"Picked Fruit: {picked_fruit}")
````

## 12. Create and import your own module

A Python module is simply a `.py` file. In this example, we create `calculator.py` and then import it.

The file is created in the notebook's current working folder.

````python
from pathlib import Path

module_code = """
def add(a, b):
    return a + b

def multiply(a, b):
    return a * b
"""

Path("calculator.py").write_text(module_code.strip() + "\n", encoding="utf-8")
print("calculator.py created!")
````

````python
import calculator

print(calculator.add(10, 5))
print(calculator.multiply(4, 3))
````

# Packages

## 13. What is a package?

A **package** is a folder used to organise related Python modules.

Example structure:

```text
my_package/
    greetings.py
    maths_tools.py
```

### Why use packages?
When a project becomes larger, packages group related code together and make it easier to maintain.

### Where are packages useful?
- Data science projects
- Web applications
- Machine learning projects
- Large Python applications

## 14. Module vs Package

| Module | Package |
|---|---|
| Usually one `.py` file | Usually a folder containing modules |
| Holds functions, classes, or variables | Organises related modules |
| Example: `calculator.py` | Example: `my_package/` |

**Easy way to remember:** A module is like one notebook; a package is like a folder holding related notebooks.
