# Python Basics: Strings, Useful Operations & Exception Handling

**Learning goals**
- Understand strings and common string operations.
- Use useful built-in operations such as `len()`, `in`, `split()`, and `join()`.
- Understand errors and handle them using `try`, `except`, `else`, and `finally`.

## 1. Repeating a string

The `*` operator repeats a string a specified number of times.

```python
print("Hi! " * 3)
print("-" * 20)
```

## 2. Finding the length with `len()`

`len()` returns the number of characters in a string. Spaces and punctuation count too.

```python
word = "Python"
sentence = "Hello   world!"

print(len(word))       # 6
print(len(sentence))   # 12, including the space and !
```

## 3. Accessing characters: indexing

Each character has a position called an **index**. Python starts counting at `0`.

For `"PYTHON"`:
- index `0` is `P`
- index `1` is `Y`
- index `-1` is the last character, `N`

```python
language = "PYTHON"

print(language[0])
print(language[1])
print(language[-1])
print(language[-2])
```

## 4. Taking part of a string: slicing

Slicing extracts a section of a string using `text[start:stop]`.

- `start` is included.
- `stop` is not included.
- Leaving out `start` or `stop` uses the beginning or end.

```python
text = "Python Programming"

print(text[0:6])   # Python
print(text[7:])    # Programming
print(text[:6])    # Python
print(text[-11:])  # Programming
```

## 5. Changing letter case

These methods return a new string; they do not change the original string in place.

- `upper()` — uppercase letters
- `lower()` — lowercase letters
- `title()` — first letter of each word in uppercase
- `capitalize()` — first character uppercase, remaining letters lowercase

```python
text = "welcome TO. python"

print(text.upper())
print(text.lower())
print(text.title())
print(text.capitalize())
print("Original:", text)
```

## 6. Removing extra spaces with `strip()`

- `strip()` removes whitespace from both ends.
- `lstrip()` removes whitespace from the left.
- `rstrip()` removes whitespace from the right.

These methods do not remove spaces between words.

```python
user_input = "   hello python   "

print(user_input)
print(user_input.strip())
print(user_input.lstrip())
print(user_input.rstrip())
```

## 7. Replacing text with `replace()`

`replace(old, new)` returns a string where matching text is replaced.

```python
sentence = "I like Java"
updated_sentence = sentence.replace("Java", "Python")

print(updated_sentence)
print(sentence)  # Original string stays unchanged
```

## 8. Searching inside a string

- `in` checks whether some text exists and returns `True` or `False`.
- `find()` returns the starting index, or `-1` if not found.
- `count()` counts non-overlapping occurrences.

```python
sentence = "Python is easy. Python is useful."

print("Python" in sentence)
print("Java" in sentence)
print(sentence.find("easy"))
print(sentence.find("Java"))
print(sentence.count("Python"))
```

## 9. Splitting a string into a list

`split()` breaks a string into pieces. By default, it splits at whitespace.

```python
skills = "Python SQL Excel, Power BI"

skill_list = skills.split()
print(skill_list)
print(skill_list[0])
print(len(skill_list))
```

## 10. Joining a list into a string

`join()` combines strings using a separator. The separator goes before `.join(...)`.

Note: all list items must be strings.

```python
skills = ["Python", "SQL", "Excel"]

print(" ".join(skills))
print(" | ".join(skills))
print("".join(skills))
```

## 11. Checking what a string contains

These methods return `True` or `False`:
- `isalpha()` — letters only
- `isdigit()` — digits only
- `isalnum()` — letters and/or digits, with no spaces or punctuation
- `isspace()` — whitespace only

An empty string returns `False` for these checks.

```python
print("Python".isalpha())
print("12345".isdigit())
print("Python.123".isalnum())
print("Python 123".isalnum())  # False because of the space
print("   ".isspace())
```

## 12. Useful built-in operations: `min()`, `max()`, `sum()`, `sorted()`

These are useful when working with numbers in a list.

```python
marks = [78, 92, 65, 88, 74]

print("Smallest:", min(marks))
print("Largest:", max(marks))
print("Total:", sum(marks))
print("Sorted:", sorted(marks))
print("Count:", len(marks))
print("Average:", sum(marks) / len(marks))
```

## 13. What is an exception?

An **exception** is an error that happens while a program is running. If it is not handled, the program may stop.

Examples:
- Dividing by zero
- Converting `"hello"` to an integer
- Accessing a list index that does not exist

## 14. Handle errors using `try` and `except`

Put code that may fail inside `try`. If the expected error occurs, Python runs the matching `except` block instead of stopping at that error.

```python
try:
    age = int("hello")
    print("Age:", age)
except Exception as e:
    print(e)
```

## 15. Handling more than one type of exception

Different problems raise different exception types. Handle only the errors you expect and can respond to meaningfully.

```python
# Example A: invalid number
try:
    number = int("ten")
except ValueError:
    print("That text is not a valid integer.")

# Example B: division by zero
try:
    result = 10 / 0
except ZeroDivisionError:
    print("You cannot divide by zero.")
```

## 16. `else` and `finally`

- `else` runs when the `try` block completes without an exception.
- `finally` runs whether an exception happened or not. It is often used for cleanup.

```python
try:
    number = int("25")
except ValueError:
    print("Please enter a valid number.")
else:
    print("Conversion worked:", number)
finally:
    print("This message always runs.")
```

## 17. Safe user input

`input()` always returns text. Convert it when you need a number, and handle invalid input.

```python
user_text = "18"  # Replace this with input("Enter your age: ") to try it interactively

try:
    age = int(user_text)
    if age >= 18:
        print("You are 18 or older.")
    else:
        print("You are under 18.")
except ValueError:
    print("Please enter your age using digits, for example 18.")
```

## 18. Avoid a common mistake: overly broad `except`

Using `except:` catches almost every ordinary error, including errors you may not expect. Prefer a specific exception such as `ValueError` or `ZeroDivisionError`, so real problems are not hidden.

```python
# Recommended
try:
    quantity = int("five")
except ValueError:
    print("Quantity must be a whole number.")
```

## 26. Mini project: Safe calculator

The calculator asks for two numbers and an operator. Try changing `first_text`, `second_text`, and `operator` to test different cases. The program handles invalid numbers and division by zero.

```python
first_text = "20"
second_text = "5"
operator = "/"

try:
    first = float(first_text)
    second = float(second_text)

    if operator == "+":
        answer = first + second
    elif operator == "-":
        answer = first - second
    elif operator == "*":
        answer = first * second
    elif operator == "/":
        answer = first / second
    else:
        raise ValueError("Unsupported operator")

    print("Answer:", answer)

except ValueError as error:
    print("Input or operator error:", error)
except ZeroDivisionError:
    print("Cannot divide by zero.")
```
